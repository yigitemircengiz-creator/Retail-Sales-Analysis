```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(14, 7))
sns.set_theme(style="whitegrid")

ax = sns.barplot(data=my_data, x='Category', y='Profit', palette='magma', ci=None)

plt.title('Profitability Analysis Across Product Categories', fontsize=18, fontweight='bold', color='#2c3e50', pad=30)
plt.xlabel('Market Category', fontsize=13, fontweight='semibold')
plt.ylabel('Net Profit (USD)', fontsize=13, fontweight='semibold')

for p in ax.patches:
    ax.annotate(format(p.get_height(), '.1f'), 
                   (p.get_x() + p.get_width() / 2., p.get_height()), 
                   ha = 'center', va = 'center', 
                   xytext = (0, 9), 
                   textcoords = 'offset points',
                   fontsize=11, fontweight='bold')

sns.despine()
plt.show()
```
